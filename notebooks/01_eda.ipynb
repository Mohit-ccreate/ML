# 01 — EDA for OptionEdge
Explore Nifty OHLC + PCR + Greeks before modeling

Run: `python src/data/nse_data.py --symbol NIFTY --days 365` first

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
from pathlib import Path
sns.set(style='whitegrid')

raw = Path('../data/raw/NIFTY_ohlc_365d.csv')
proc = Path('../data/processed/NIFTY_features.csv')
if proc.exists():
    df = pd.read_csv(proc)
elif raw.exists():
    df = pd.read_csv(raw)
else:
    print('No data found. Run: python src/data/nse_data.py --symbol NIFTY')
    df = pd.DataFrame()
df.head()
```

```python
if not df.empty:
    print(df[['Close','Return','RSI_14','Vol_20']].describe().round(2))
    df['Target'].value_counts().plot(kind='bar', title='Target Distribution (-1 Down, 0 Sideways, 1 Up)')
    plt.show()
```

```python
if not df.empty and 'Close' in df.columns:
    plt.figure(figsize=(12,4))
    plt.plot(pd.to_datetime(df['Date']), df['Close'])
    plt.title('Nifty Close')
    plt.show()
    
    # RSI
    if 'RSI_14' in df.columns:
        plt.figure(figsize=(12,2))
        plt.plot(df['RSI_14'])
        plt.axhline(70, color='red', ls='--')
        plt.axhline(30, color='green', ls='--')
        plt.title('RSI 14')
        plt.show()
```

## PCR vs Next Day Move (if you have option chain time series)
Add your daily PCR time series here to see edge.

```python
if not df.empty and 'Target_return' in df.columns:
    print('Correlation with next return:')
    for col in ['RSI_14','MACD','BB_pct','Vol_20','Return_lag_1']:
        if col in df.columns:
            print(col, df[col].corr(df['Target_return']).round(3))
```
